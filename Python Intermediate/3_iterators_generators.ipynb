## 3.1 Iterator Protocol — `__iter__` / `__next__` 🔧

```python
# Under the hood: what 'for' actually does
nums = [10, 20, 30]
it = iter(nums)         # calls nums.__iter__()
print(next(it))         # 10  — calls it.__next__()
print(next(it))         # 20
print(next(it))         # 30
# next(it)              # StopIteration!
```

```python
# Custom iterator
class Countdown:
    def __init__(self, start):
        self.start = start
    def __iter__(self):
        self.current = self.start
        return self
    def __next__(self):
        if self.current < 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val

for n in Countdown(5):
    print(n, end=" ")  # 5 4 3 2 1 0
print("🚀")
```

## 3.2 Generators with `yield` ⚡

A **generator** is a function that **pauses** and **resumes**. Uses `yield` instead of `return`.

```python
def count_up(start, end):
    """Generator: yields numbers from start to end."""
    current = start
    while current <= end:
        yield current     # pauses here, gives value
        current += 1      # resumes on next call

gen = count_up(1, 5)
print(type(gen))       # <class 'generator'>
print(next(gen))       # 1
print(next(gen))       # 2
print(list(gen))       # [3, 4, 5] — remaining values

# Much simpler than writing a class with __iter__/__next__!
```

```python
# Why generators? MEMORY EFFICIENCY!
import sys

# List: stores ALL 1M numbers in memory
big_list = [x**2 for x in range(1_000_000)]
print(f"List size: {sys.getsizeof(big_list):,} bytes")

# Generator: computes ONE at a time
big_gen = (x**2 for x in range(1_000_000))
print(f"Generator size: {sys.getsizeof(big_gen):,} bytes")

# Same result, fraction of memory!
print(f"Sum matches: {sum([x**2 for x in range(100)]) == sum(x**2 for x in range(100))}")
```

### ✏️ Exercise 4 — Generator Pipeline 🔗

```python
# Chain generators together like UNIX pipes!

def numbers(n):
    """Generate 1 to n."""
    for i in range(1, n + 1):
        yield i

def squared(gen):
    """Square each value from a generator."""
    for val in gen:
        yield val ** 2

def only_even(gen):
    """Filter: keep only even values."""
    for val in gen:
        if val % 2 == 0:
            yield val

# Pipeline: numbers → square → filter even
pipeline = only_even(squared(numbers(10)))
print(list(pipeline))  # [4, 16, 36, 64, 100]

# Each step is lazy — nothing computed until we consume!
```

## 3.3 `itertools` — Power Tools for Iteration 🧰

```python
import itertools as it

# chain — concatenate iterables
print(list(it.chain([1,2], [3,4], [5,6])))  # [1,2,3,4,5,6]

# islice — slice any iterable (even infinite ones!)
print(list(it.islice(range(100), 5, 15, 2)))  # [5,7,9,11,13]

# count — infinite counter
print(list(it.islice(it.count(10, 5), 5)))  # [10,15,20,25,30]

# cycle — repeat infinitely
print(list(it.islice(it.cycle(["🔴","🟢","🔵"]), 8)))

# repeat
print(list(it.repeat("Hello", 3)))  # ['Hello', 'Hello', 'Hello']
```

```python
# Combinatorics
print("product:", list(it.product("AB", "12")))   # [('A','1'),('A','2'),('B','1'),('B','2')]
print("combos:",  list(it.combinations("ABCD", 2)))  # pairs
print("perms:",   list(it.permutations("ABC", 2)))   # ordered pairs
```

```python
# groupby — group consecutive elements
data = [("fruit", "apple"), ("fruit", "banana"), ("veg", "carrot"), ("veg", "pea"), ("fruit", "mango")]
data.sort(key=lambda x: x[0])  # must sort first!

for key, group in it.groupby(data, key=lambda x: x[0]):
    items = [item[1] for item in group]
    print(f"  {key}: {items}")
```

## 3.4 `collections` Module — Power Data Structures 💪

```python
from collections import Counter

# Counter — count occurrences
text = "abracadabra"
c = Counter(text)
print(f"Counts: {c}")
print(f"Most common: {c.most_common(3)}")

words = "the cat sat on the mat the cat".split()
wc = Counter(words)
print(f"Word counts: {wc}")

# Arithmetic with Counters!
c1 = Counter(a=3, b=1)
c2 = Counter(a=1, b=2)
print(f"c1 + c2 = {c1 + c2}")
print(f"c1 - c2 = {c1 - c2}")  # only positive counts
```

```python
from collections import defaultdict

# defaultdict — never get KeyError!
# Regular dict:
# d = {}; d["key"] += 1  → KeyError!

dd = defaultdict(int)        # default value = 0
for word in "hello world hello python world hello".split():
    dd[word] += 1
print(dict(dd))  # {'hello': 3, 'world': 2, 'python': 1}

# Group items
students = [("Math", "Anita"), ("Science", "Rahul"), ("Math", "Sneha"), ("Science", "Priya")]
by_subject = defaultdict(list)
for subj, name in students:
    by_subject[subj].append(name)
print(dict(by_subject))
```

```python
from collections import namedtuple, deque

# namedtuple — lightweight immutable class
Point = namedtuple("Point", ["x", "y"])
p = Point(3, 4)
print(f"Point: {p}, x={p.x}, y={p.y}")

# deque — double-ended queue (fast append/pop from both ends)
dq = deque([1, 2, 3], maxlen=5)
dq.append(4)       # right
dq.appendleft(0)   # left
print(f"deque: {dq}")
dq.rotate(2)       # rotate right
print(f"rotated: {dq}")
```

### ✏️ Exercise 10 — deque: Browser History 🌐

```python
from collections import deque

class BrowserHistory:
    def __init__(self, max_history=10):
        self._history = deque(maxlen=max_history)
        self._current = -1
    
    def visit(self, url):
        # Remove forward history when visiting new page
        while len(self._history) > self._current + 1:
            self._history.pop()
        self._history.append(url)
        self._current = len(self._history) - 1
        print(f"  → {url}")
    
    def back(self):
        if self._current > 0:
            self._current -= 1
            print(f"  ← Back: {self._history[self._current]}")
        else:
            print("  ← No more history")
    
    def forward(self):
        if self._current < len(self._history) - 1:
            self._current += 1
            print(f"  → Forward: {self._history[self._current]}")
        else:
            print("  → No forward pages")
    
    def show(self):
        for i, url in enumerate(self._history):
            marker = " 👈" if i == self._current else ""
            print(f"  {i+1}. {url}{marker}")

browser = BrowserHistory()
browser.visit("google.com")
browser.visit("python.org")
browser.visit("github.com")
browser.back()
browser.back()
browser.forward()
browser.show()
```

### ✏️ Exercise 11 — Generator: File Line Streamer 📄

```python
# TODO: Create a generator that reads a file line by line,
# strips whitespace, and skips empty lines

def read_lines(filepath):
    """Generator that yields non-empty, stripped lines from a file."""
    try:
        with open(filepath, "r") as f:
            for line in f:
                stripped = line.strip()
                if stripped:  # skip empty lines
                    yield stripped
    except FileNotFoundError:
        print(f"❌ File not found: {filepath}")

# Test with the sample file from basics
for i, line in enumerate(read_lines("../data/sample_text.txt"), 1):
    print(f"  {i:>2}: {line[:60]}")
```

### ✏️ Exercise 12 — `yield from` — Delegating Generators 🔗

```python
# yield from delegates to another iterable/generator

def flatten(nested_list):
    """Recursively flatten any nested list."""
    for item in nested_list:
        if isinstance(item, list):
            yield from flatten(item)  # delegate to recursive call
        else:
            yield item

nested = [1, [2, 3], [4, [5, 6]], [[7, 8], 9]]
print(list(flatten(nested)))  # [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Also works for chaining generators
def multi_range(*ranges):
    for r in ranges:
        yield from r

print(list(multi_range(range(3), range(10, 13), range(20, 22))))
# [0, 1, 2, 10, 11, 12, 20, 21]
```

### ✏️ Exercise 13 — Counter + defaultdict: Log Analyzer 📋

```python
from collections import Counter, defaultdict

# Simulated log entries
logs = [
    "2024-01-15 ERROR Database connection failed",
    "2024-01-15 INFO User login: anita",
    "2024-01-15 WARNING Disk usage 85%",
    "2024-01-15 ERROR Timeout on API call",
    "2024-01-16 INFO User login: rahul",
    "2024-01-16 ERROR Database connection failed",
    "2024-01-16 INFO User logout: anita",
    "2024-01-16 WARNING Memory usage high",
    "2024-01-17 ERROR Timeout on API call",
    "2024-01-17 INFO Server restart",
]

# Parse logs
def parse_log(line):
    parts = line.split(" ", 2)
    return {"date": parts[0], "level": parts[1], "message": parts[2]}

parsed = [parse_log(log) for log in logs]

# Count by level
level_counts = Counter(log["level"] for log in parsed)
print("📊 Log Level Counts:")
for level, count in level_counts.most_common():
    print(f"  {level:<10} {count}")

# Group errors by message
errors = defaultdict(int)
for log in parsed:
    if log["level"] == "ERROR":
        errors[log["message"]] += 1

print("\n❌ Error Summary:")
for msg, count in sorted(errors.items(), key=lambda x: -x[1]):
    print(f"  {msg}: {count}x")
```

### ✏️ Exercise 14 — Sliding Window with deque 📈

```python
from collections import deque

def moving_average(data, window_size):
    """Generator that yields moving averages."""
    window = deque(maxlen=window_size)
    for value in data:
        window.append(value)
        if len(window) == window_size:
            yield sum(window) / window_size

# Stock prices
prices = [100, 102, 104, 103, 105, 108, 107, 110, 112, 115]
ma3 = list(moving_average(prices, 3))
ma5 = list(moving_average(prices, 5))

print("📈 Stock Price Analysis:")
print(f"Prices: {prices}")
print(f"MA(3):  {[f'{x:.1f}' for x in ma3]}")
print(f"MA(5):  {[f'{x:.1f}' for x in ma5]}")
```

## 🎯 Exercise 15 — Mini-Project: Log Stream Processor 📊

```python
import itertools as it
from collections import Counter, defaultdict
import random
from datetime import datetime, timedelta

# === Generate fake log data ===
def generate_logs(n=50):
    levels = ["INFO"] * 5 + ["WARNING"] * 2 + ["ERROR"]
    messages = {
        "INFO": ["User login", "Page viewed", "API call", "Cache hit"],
        "WARNING": ["Slow query", "High memory", "Rate limit near"],
        "ERROR": ["DB timeout", "Auth failed", "500 Internal"],
    }
    base = datetime(2024, 1, 1)
    for i in range(n):
        level = random.choice(levels)
        msg = random.choice(messages[level])
        ts = base + timedelta(hours=i)
        yield f"{ts.strftime('%Y-%m-%d %H:%M')} [{level}] {msg}"

# === Generator pipeline ===
def parse(logs):
    for log in logs:
        parts = log.split(" ", 2)
        ts = parts[0] + " " + parts[1]
        level = parts[2].split("]")[0].replace("[", "")
        msg = parts[2].split("] ")[1]
        yield {"timestamp": ts, "level": level, "message": msg}

def filter_level(logs, level):
    for log in logs:
        if log["level"] == level:
            yield log

# === Process ===
random.seed(42)
raw_logs = list(generate_logs(100))
parsed = list(parse(raw_logs))

# Stats
level_counts = Counter(log["level"] for log in parsed)
error_msgs = Counter(log["message"] for log in filter_level(parse(raw_logs), "ERROR"))

print("═" * 50)
print(f"{'📊 LOG STREAM ANALYSIS':^50}")
print("═" * 50)
print(f"  Total entries: {len(parsed)}")
print(f"\n  📈 By Level:")
for level, count in level_counts.most_common():
    pct = count / len(parsed) * 100
    bar = "█" * int(pct / 2)
    print(f"    {level:<10} {count:>3} ({pct:>5.1f}%) {bar}")
print(f"\n  ❌ Top Errors:")
for msg, count in error_msgs.most_common(5):
    print(f"    {msg:<20} {count:>3}x")
print("═" * 50)
```
