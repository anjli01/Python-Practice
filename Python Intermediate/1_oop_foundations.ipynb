## 1.2 Classes & Objects — The Basics 📦

A **class** = blueprint. An **object** = instance built from that blueprint.


````python
class Dog:
    def __init__(self, name, breed, age):
        self.name = name
        self.breed = breed
        self.age = age
    
    def bark(self):
        return f"{self.name} says: Woof! 🐕"
    
    def describe(self):
        return f"{self.name} is a {self.age}-year-old {self.breed}"

dog1 = Dog("Buddy", "Golden Retriever", 3)
dog2 = Dog("Luna", "German Shepherd", 5)

print(dog1.bark())
print(dog2.describe())
print(f"Type: {type(dog1)}")
print(f"Same object? {dog1 is dog2}")
````

````python
# 'self' = reference to the CURRENT object
class Counter:
    def __init__(self):
        self.count = 0
    def increment(self):
        self.count += 1
        return self.count

c1, c2 = Counter(), Counter()
c1.increment(); c1.increment(); c2.increment()
print(f"c1: {c1.count}, c2: {c2.count}")  # 2, 1 — independent!
````

## 1.4 Class Variables vs Instance Variables ⚖️

| | Instance Variable | Class Variable |
|-|-------------------|----------------|
| Belongs to | Each object | The class itself |
| Defined in | `__init__` via `self.x` | Directly in class body |
| Shared? | ❌ Own copy | ✅ Shared across all |

````python
class Employee:
    company = "TechCorp"    # class variable
    count = 0
    
    def __init__(self, name, salary):
        self.name = name           # instance variable
        self.salary = salary
        Employee.count += 1        # modify via CLASS name

e1 = Employee("Anita", 95000)
e2 = Employee("Rahul", 80000)
print(f"Count: {Employee.count}")   # 2
Employee.company = "MegaCorp"        # affects ALL
print(f"{e1.name}: {e1.company}")   # MegaCorp
````

## 1.5 `@staticmethod` and `@classmethod` 🏷️

| Decorator | Gets `self`? | Gets `cls`? | Use Case |
|-----------|:-----------:|:-----------:|----------|
| (regular) | ✅ | ❌ | Access/modify instance |
| `@classmethod` | ❌ | ✅ | Alternative constructors |
| `@staticmethod` | ❌ | ❌ | Utility functions |

````python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius
    
    def to_fahrenheit(self):            # instance method
        return self.celsius * 9/5 + 32
    
    @classmethod
    def from_fahrenheit(cls, f):        # alternative constructor
        return cls((f - 32) * 5/9)
    
    @staticmethod
    def is_boiling(c):                  # utility — no self/cls
        return c >= 100

t1 = Temperature(100)
print(f"{t1.celsius}°C = {t1.to_fahrenheit()}°F")
t2 = Temperature.from_fahrenheit(98.6)  # classmethod
print(f"Body temp: {t2.celsius:.1f}°C")
print(f"Boiling? {Temperature.is_boiling(100)}")  # staticmethod
````

## 1.6 `__str__` and `__repr__` — Readable Printing 🖨️

| Method | Purpose | Called by |
|--------|---------|----------|
| `__str__` | Human-readable | `print()`, f-strings |
| `__repr__` | Developer-readable | `repr()`, debugger, lists |

````python
class Book:
    def __init__(self, title, author, year):
        self.title, self.author, self.year = title, author, year
    
    def __str__(self):
        return f"📚 '{self.title}' by {self.author} ({self.year})"
    
    def __repr__(self):
        return f"Book('{self.title}', '{self.author}', {self.year})"

b = Book("The Alchemist", "Paulo Coelho", 1988)
print(b)          # __str__
print(repr(b))    # __repr__
print([b])        # __repr__ in list
````

## 1.7 Encapsulation — Public, Protected, Private 🔒

| Convention | Prefix | Meaning |
|-----------|--------|--------|
| Public | `name` | Anyone can access |
| Protected | `_name` | Internal use hint |
| Name-mangled | `__name` | Hard to access from outside |

````python
class BankAccountDemo:
    def __init__(self, owner, balance):
        self.owner = owner             # public
        self._type = "Savings"         # protected
        self.__balance = balance        # name-mangled
    
    def get_balance(self): return self.__balance

acc = BankAccountDemo("Priya", 10000)
print(acc.owner)           # ✅
print(acc._type)           # ✅ but shouldn't
# print(acc.__balance)     # ❌ AttributeError!
print(acc.get_balance())   # ✅ via method
````

## 1.8 `@property` — Getters & Setters the Pythonic Way 🏠

````python
class Circle:
    def __init__(self, radius):
        self._radius = radius
    
    @property
    def radius(self): return self._radius
    
    @radius.setter
    def radius(self, value):
        if value <= 0: raise ValueError("Must be positive!")
        self._radius = value
    
    @property
    def area(self): return 3.14159 * self._radius ** 2

c = Circle(5)
print(f"r={c.radius}, area={c.area:.2f}")
c.radius = 10
print(f"r={c.radius}, area={c.area:.2f}")
# c.radius = -5  → ValueError!
````

### ✏️ Exercise 14 — Inventory Management 📦

````python
class Item:
    _next_id = 1
    def __init__(self, name, cat, price, qty):
        self.id = Item._next_id; Item._next_id += 1
        self.name, self.cat = name, cat
        self._price, self._qty = price, qty
    
    @property
    def price(self): return self._price
    @price.setter
    def price(self, v):
        if v < 0: raise ValueError("Negative!")
        self._price = v
    
    @property
    def qty(self): return self._qty
    @qty.setter
    def qty(self, v):
        if v < 0: raise ValueError("Negative!")
        self._qty = v
    
    @property
    def value(self): return self._price * self._qty
    @property
    def in_stock(self): return self._qty > 0
    
    def __str__(self):
        s = "✅" if self.in_stock else "❌"
        return f"[{self.id:>3}] {self.name:<15} {self.cat:<12} ₹{self.price:>8,} ×{self.qty:>4} = ₹{self.value:>10,} {s}"

class Inventory:
    def __init__(self): self._items = []
    def add(self, item): self._items.append(item)
    def sell(self, id, n):
        for i in self._items:
            if i.id == id:
                if i.qty >= n: i.qty -= n; return f"✅ Sold {n}× {i.name}"
                return f"❌ Only {i.qty} left!"
        return "❌ Not found"
    @property
    def total_value(self): return sum(i.value for i in self._items)
    def report(self):
        print("═" * 72)
        print(f"{'📦 INVENTORY':^72}")
        print("═" * 72)
        for i in self._items: print(f"  {i}")
        print(f"{'─'*72}\n  Total: ₹{self.total_value:,}")
        oos = [i.name for i in self._items if not i.in_stock]
        if oos: print(f"  ⚠️ Out: {', '.join(oos)}")

inv = Inventory()
inv.add(Item("Laptop", "Electronics", 75000, 10))
inv.add(Item("Mouse", "Electronics", 500, 50))
inv.add(Item("Chair", "Furniture", 12000, 5))
inv.report()
print(inv.sell(1, 3))
inv.report()
````

## 🎯 Exercise 15 — Mini-Project: Bank Account System 🏦

````python
from datetime import datetime

class BankAccount:
    _next = 1000
    bank_name = "Python National Bank"
    interest_rate = 0.04
    
    def __init__(self, holder, initial=0):
        BankAccount._next += 1
        self._acct = f"PNB{BankAccount._next}"
        self._holder = holder
        self.__bal = initial
        self._txns = []
        self._log("OPEN", initial, "Account opened")
    
    def _log(self, t, amt, note=""):
        self._txns.append({"time": datetime.now().strftime("%H:%M:%S"),
                           "type": t, "amt": amt, "bal": self.__bal, "note": note})
    
    @property
    def balance(self): return self.__bal
    
    def deposit(self, amt):
        if amt <= 0: return "❌ Must be positive!"
        self.__bal += amt
        self._log("CR", amt, "Deposit")
        return f"✅ +₹{amt:,} → ₹{self.__bal:,}"
    
    def withdraw(self, amt):
        if amt <= 0: return "❌ Must be positive!"
        if amt > self.__bal: return f"❌ Insufficient! ₹{self.__bal:,}"
        self.__bal -= amt
        self._log("DR", amt, "Withdrawal")
        return f"✅ -₹{amt:,} → ₹{self.__bal:,}"
    
    def apply_interest(self):
        interest = self.__bal * self.interest_rate
        self.__bal += interest
        self._log("INT", interest, f"{self.interest_rate*100}%")
        return f"✅ Interest ₹{interest:,.2f} → ₹{self.__bal:,.2f}"
    
    def statement(self, n=5):
        print(f"\n{'═'*60}")
        print(f"  🏦 {self.bank_name} | {self._acct} | {self._holder}")
        print(f"  Balance: ₹{self.__bal:,.2f}")
        print(f"{'─'*60}")
        for t in self._txns[-n:]:
            print(f"  {t['time']} {t['type']:<6} ₹{t['amt']:>9,.0f}  bal ₹{t['bal']:>9,.0f}  {t['note']}")
        print(f"{'═'*60}")
    
    def __str__(self):
        return f"🏦 {self._holder} ({self._acct}) — ₹{self.__bal:,.2f}"

# Demo
a1 = BankAccount("Priya", 50000)
print(a1)
print(a1.deposit(15000))
print(a1.withdraw(5000))
print(a1.apply_interest())
a1.statement()
````
