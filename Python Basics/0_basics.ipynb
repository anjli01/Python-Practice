## 🔤 Section 3 — String Operations

Strings are text. Python gives you tons of tools to work with them!

````python
greeting = "Hello, World!"

# Length
print(len(greeting))          # 13 characters

# Uppercase & Lowercase
print(greeting.upper())       # HELLO, WORLD!
print(greeting.lower())       # hello, world!

# Replace
print(greeting.replace("World", "Python"))  # Hello, Python!

# Indexing (positions start at 0!)
print(greeting[0])            # H
print(greeting[7])            # W

# Slicing — get a piece of the string
print(greeting[0:5])          # Hello
print(greeting[7:])           # World!

# Concatenation — joining strings with +
first = "Good"
second = " Morning!"
print(first + second)         # Good Morning!
````

## 🔀 Section 6 — Conditionals (if / elif / else)

Conditionals let Python **make decisions**.

```python
if condition:
    do this
elif another_condition:
    do this instead
else:
    do this if nothing above was true
```

⚠️ **Indentation matters!** The code inside `if/elif/else` must be indented (4 spaces).

````python
temperature = 35

if temperature > 30:
    print("It's hot! 🥵 Stay hydrated.")
elif temperature > 20:
    print("Nice weather! 😊")
else:
    print("It's cold! 🥶 Wear a jacket.")
````

````python
# You can combine conditions with 'and', 'or', 'not'
age = 25
has_license = False

if age >= 18 and has_license:
    print("You can drive! 🚗")
elif age >= 18 and not has_license:
    print("You're old enough, but get a license first!")
else:
    print("You're too young to drive.")
````

## 📋 Section 7 — Lists

A **list** stores multiple values in a single variable. Lists use square brackets `[]`.

````python
# Creating a list
fruits = ["Mango", "Apple", "Banana", "Grapes"]
print(fruits)

# Accessing items (index starts at 0!)
print(fruits[0])     # Mango
print(fruits[-1])    # Grapes (last item)

# Length
print(len(fruits))   # 4

# Add an item
fruits.append("Papaya")
print(fruits)        # ['Mango', 'Apple', 'Banana', 'Grapes', 'Papaya']

# Remove an item
fruits.remove("Banana")
print(fruits)        # ['Mango', 'Apple', 'Grapes', 'Papaya']

# Check if item exists
print("Mango" in fruits)     # True
print("Cherry" in fruits)    # False

# Slicing
print(fruits[1:3])   # ['Apple', 'Grapes']
````

## 🔁 Section 8 — Loops

Loops let you **repeat** actions. Python has two types:

**`for` loop** — repeat for each item in a list or range  
**`while` loop** — repeat as long as a condition is true

````python
# for loop — go through each item in a list
foods = ["Pizza", "Pasta", "Sushi", "Tacos"]

for food in foods:
    print(f"I love {food}! 😋")
````

````python
# range() — generate a sequence of numbers
for i in range(5):         # 0, 1, 2, 3, 4
    print(f"Count: {i}")

print("---")

for i in range(1, 6):      # 1, 2, 3, 4, 5
    print(f"Step {i}")
````

````python
# while loop — repeat as long as condition is True
countdown = 5

while countdown > 0:
    print(f"{countdown}...")
    countdown = countdown - 1

print("🚀 Blast off!")
````

## 🛠️ Section 9 — Functions

A **function** is a reusable block of code. You **define** it once, then **call** it whenever you need it.

```python
def function_name(parameter1, parameter2):
    # do something
    return result
```

````python
# A simple function
def greet(name):
    return f"Hello, {name}! 👋"

# Calling the function
message = greet("Priya")
print(message)

print(greet("Vishal"))
````

````python
# Function with default parameter
def make_coffee(size="medium", milk=True):
    drink = f"One {size} coffee"
    if milk:
        drink += " with milk"
    else:
        drink += ", black"
    return drink + " ☕"

print(make_coffee())                    # One medium coffee with milk ☕
print(make_coffee("large"))             # One large coffee with milk ☕
print(make_coffee("small", milk=False)) # One small coffee, black ☕
````

    if total > 5000:
        discount_percent = 15
    elif total > 3000:
        discount_percent = 10
    else:
        discount_percent = 0
    discount_amount = total * discount_percent / 100
    final_total = total - discount_amount
    return final_total, discount_percent, discount_amount
