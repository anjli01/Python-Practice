## 6.2 Parameters Deep Dive 🔧

| Type | Example | Description |
|------|---------|-------------|
| Positional | `def f(a, b)` | Required, in order |
| Default | `def f(a, b=10)` | Optional, has fallback |
| Keyword | `f(b=20, a=10)` | Named, any order |
| `*args` | `def f(*args)` | Variable positional args (tuple) |
| `**kwargs` | `def f(**kwargs)` | Variable keyword args (dict) |

````python
# Default parameters
def make_coffee(size="medium", sugar=2, milk=True):
    drink = f"{size.title()} coffee, {sugar} sugar"
    drink += ", with milk" if milk else ", black"
    return drink

print(make_coffee())                          # all defaults
print(make_coffee("large"))                    # override size
print(make_coffee(sugar=0, milk=False))        # keyword args
print(make_coffee("small", milk=False))        # mix positional + keyword
````

````python
# *args — accept any number of positional arguments
def add_all(*numbers):
    """Sum any number of arguments."""
    print(f"  Received: {numbers} (type: {type(numbers).__name__})")
    return sum(numbers)

print(add_all(1, 2))                # 3
print(add_all(1, 2, 3, 4, 5))       # 15
print(add_all(10))                   # 10
````

````python
# **kwargs — accept any number of keyword arguments
def build_profile(**details):
    """Build a user profile from keyword arguments."""
    print(f"  Received: {details} (type: {type(details).__name__})")
    profile = ""
    for key, value in details.items():
        profile += f"  {key}: {value}\n"
    return profile

print(build_profile(name="Priya", age=28, city="Mumbai"))
print(build_profile(username="priya123", role="admin"))
````

## 6.3 Scope & Namespaces 🔭

````python
# Global vs Local scope
x = 10          # global variable

def show_x():
    x = 20      # local variable (shadows global)
    print(f"Inside function: x = {x}")  # 20

show_x()
print(f"Outside function: x = {x}")     # 10 (unchanged!)

# To modify global from inside a function:
counter = 0
def increment():
    global counter
    counter += 1

increment()
increment()
print(f"Counter: {counter}")  # 2
````

## 6.4 Lambda Functions ⚡

A **lambda** is a small anonymous function defined in one line.

```python
lambda parameters: expression
```

````python
# Lambda vs regular function
def square(x): return x ** 2
square_lambda = lambda x: x ** 2

print(square(5))           # 25
print(square_lambda(5))    # 25

# Lambdas are great as sort keys
students = [("Anita", 88), ("Rahul", 72), ("Sneha", 95), ("Vishal", 68)]

# Sort by score (second element)
by_score = sorted(students, key=lambda s: s[1], reverse=True)
print(f"By score: {by_score}")

# Sort strings by length
words = ["banana", "pie", "strawberry", "kiwi", "fig"]
by_length = sorted(words, key=lambda w: len(w))
print(f"By length: {by_length}")
````

## 6.5 `map()`, `filter()`, `zip()` 🗺️

Functional programming tools that transform data without explicit loops.

````python
# map() — apply a function to every element
nums = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x**2, nums))
print(f"Squares: {squares}")   # [1, 4, 9, 16, 25]

# filter() — keep only elements that pass a test
nums = range(1, 21)
evens = list(filter(lambda x: x % 2 == 0, nums))
print(f"Evens: {evens}")       # [2, 4, 6, ..., 20]

# Combining map + filter
# Get squares of even numbers
result = list(map(lambda x: x**2, filter(lambda x: x % 2 == 0, range(1, 11))))
print(f"Even squares: {result}")  # [4, 16, 36, 64, 100]
````

## 6.6 Recursion 🔄

A function that **calls itself**. Every recursive function needs:
1. **Base case** — when to stop
2. **Recursive case** — call itself with a smaller problem

````python
# Countdown
def countdown(n):
    if n <= 0:              # base case
        print("🚀 Liftoff!")
    else:
        print(f"{n}...")
        countdown(n - 1)    # recursive case

countdown(5)
````

````python
# Factorial: n! = n × (n-1) × (n-2) × ... × 1
def factorial(n):
    if n <= 1:          # base case
        return 1
    return n * factorial(n - 1)  # recursive case

for i in range(1, 8):
    print(f"{i}! = {factorial(i)}")
````

## 🎯 Exercise 15 — Mini-Project: Quiz Game Engine 🎮

````python
import random

# =============================================
#    🎯 MINI-PROJECT: Quiz Game Engine
# =============================================

def create_quiz():
    """Return a list of quiz questions."""
    return [
        {"question": "What is the capital of France?",
         "options": ["London", "Paris", "Berlin", "Madrid"],
         "answer": 2},
        {"question": "Which planet is known as the Red Planet?",
         "options": ["Venus", "Jupiter", "Mars", "Saturn"],
         "answer": 3},
        {"question": "What is 2 ** 10?",
         "options": ["100", "512", "1024", "2048"],
         "answer": 3},
        {"question": "Which data type is immutable in Python?",
         "options": ["List", "Dictionary", "Set", "Tuple"],
         "answer": 4},
        {"question": "What does 'len' stand for?",
         "options": ["Length", "Lengthy", "Lender", "Lens"],
         "answer": 1},
    ]

def ask_question(q, num):
    """Display a question and return True if answered correctly."""
    print(f"\n❓ Q{num}: {q['question']}")
    for i, opt in enumerate(q["options"], 1):
        print(f"   {i}. {opt}")
    
    while True:
        try:
            answer = int(input("   Your answer (1-4): "))
            if 1 <= answer <= 4:
                break
            print("   Please enter 1-4!")
        except ValueError:
            print("   Please enter a number!")
    
    correct = answer == q["answer"]
    if correct:
        print("   ✅ Correct!")
    else:
        right = q["options"][q["answer"] - 1]
        print(f"   ❌ Wrong! The answer was: {right}")
    return correct

def calculate_grade(score, total):
    """Return a grade based on percentage."""
    pct = score / total * 100
    if pct >= 90: return "A ⭐"
    elif pct >= 70: return "B 👍"
    elif pct >= 50: return "C 💪"
    else: return "F 📚"

# --- Run the quiz ---
print("🎮 PYTHON QUIZ GAME")
print("=" * 35)

questions = create_quiz()
random.shuffle(questions)

score = 0
for i, q in enumerate(questions, 1):
    if ask_question(q, i):
        score += 1

print("\n" + "=" * 35)
print(f"📊 RESULTS: {score}/{len(questions)}")
print(f"📈 Grade: {calculate_grade(score, len(questions))}")
print("=" * 35)
````
