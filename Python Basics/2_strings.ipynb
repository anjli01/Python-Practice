## 2.1 String Creation 📝

Strings are text enclosed in quotes. Python gives you multiple ways to create them.

````python
# Single quotes and double quotes — both work the same!
name1 = 'Priya'
name2 = "Priya"
print(name1 == name2)    # True — they're identical

# When your string contains quotes:
quote1 = "She said, 'Hello!'"
quote2 = 'He said, "Goodbye!"'
print(quote1)
print(quote2)
````

````python
# Triple quotes — for multi-line strings
address = """123 MG Road
Apartment 4B
Mumbai, Maharashtra
India - 400001"""
print(address)
````

````python
# === Escape Characters ===
# Special characters that start with \

print("Line 1\nLine 2")          # \n = newline
print("Column1\tColumn2")         # \t = tab
print("She said \"Hello!\"")      # \" = literal quote
print("Backslash: \\")            # \\ = literal backslash
print("\u2764")                   # Unicode: ❤

print("\n--- Escape Character Reference ---")
print("\\n  → New line")
print("\\t  → Tab")
print("\\\" → Double quote")
print("\\\\ → Backslash")
````

````python
# Raw strings — ignore escape characters (useful for file paths!)
normal = "C:\\Users\\name\\Documents"   # need to escape every \
raw = r"C:\Users\name\Documents"         # r prefix = raw string
print(normal)
print(raw)
print(normal == raw)                      # True — same result!
````

## 2.2 Indexing & Slicing ✂️

Every character in a string has a **position** (index). Indexing starts at **0**.

```
String:   P  y  t  h  o  n
Index:    0  1  2  3  4  5
Negative:-6 -5 -4 -3 -2 -1
```

**Slicing** extracts a portion: `string[start:stop:step]`
- `start` = where to begin (inclusive)
- `stop` = where to end (exclusive)
- `step` = jump size (default 1)

````python
text = "Python Programming"

# === Indexing ===
print(text[0])     # P (first character)
print(text[6])     # ' ' (space — it's a character too!)
print(text[-1])    # g (last character)
print(text[-4])    # m (4th from end)
````

````python
text = "Python Programming"

# === Slicing ===
print(text[0:6])     # Python     (positions 0,1,2,3,4,5 — stop is EXCLUDED)
print(text[7:])      # Programming (from 7 to end)
print(text[:6])      # Python     (from start to 6)
print(text[-11:])    # Programming (last 11 characters)

# With step
print(text[::2])     # Pto rgamn  (every 2nd character)
print(text[0:6:2])   # Pto        (every 2nd in "Python")

# Reverse a string!
print(text[::-1])    # gnimmargorP nohtyP
````

## 2.3 String Methods 🧰

Strings come with powerful built-in **methods** (functions attached to strings).

⚠️ **Strings are immutable** — methods return a NEW string; the original is unchanged.

````python
msg = "  Hello, World!  "

# === Case methods ===
print(msg.upper())          # '  HELLO, WORLD!  '
print(msg.lower())          # '  hello, world!  '
print(msg.title())          # '  Hello, World!  '
print(msg.capitalize())     # '  hello, world!  ' (only first char uppercase)
print(msg.swapcase())       # '  hELLO, wORLD!  '
````

````python
msg = "  Hello, World!  "

# === Stripping whitespace ===
print(f"[{msg}]")              # [  Hello, World!  ] (spaces visible)
print(f"[{msg.strip()}]")      # [Hello, World!]     (both sides stripped)
print(f"[{msg.lstrip()}]")     # [Hello, World!  ]   (left side only)
print(f"[{msg.rstrip()}]")     # [  Hello, World!]   (right side only)

# Strip specific characters
url = "###www.example.com###"
print(url.strip("#"))           # www.example.com
````

````python
# === Split & Join ===
sentence = "Python is amazing and powerful"

# split() — breaks a string into a list of words
words = sentence.split()            # splits on spaces by default
print(words)  # ['Python', 'is', 'amazing', 'and', 'powerful']

# split with custom delimiter
csv_data = "apple,banana,cherry,date"
fruits = csv_data.split(",")
print(fruits)  # ['apple', 'banana', 'cherry', 'date']

# join() — the opposite: joins a list into a string
print(" ".join(words))       # Python is amazing and powerful
print(", ".join(fruits))     # apple, banana, cherry, date
print(" → ".join(fruits))    # apple → banana → cherry → date
````

````python
# === Search methods ===
text = "I love Python. Python is great. I use Python daily."

print(text.find("Python"))       # 7  (first occurrence, index)
print(text.find("Java"))         # -1 (not found)
print(text.rfind("Python"))      # 41 (LAST occurrence)
print(text.count("Python"))      # 3  (how many times it appears)
print(text.startswith("I love")) # True
print(text.endswith("daily."))   # True
print("Python" in text)          # True (membership test)
````

````python
# === Replace ===
text = "I love cats. cats are amazing. I have 3 cats."

print(text.replace("cats", "dogs"))        # Replace ALL occurrences
print(text.replace("cats", "dogs", 1))     # Replace only FIRST occurrence

# === Check methods ===
print("hello123".isalnum())    # True  (letters + numbers only)
print("hello".isalpha())       # True  (letters only)
print("12345".isdigit())       # True  (digits only)
print("hello".islower())       # True
print("HELLO".isupper())       # True
print("  \t\n".isspace())      # True  (whitespace only)
````

## 2.4 String Formatting 🎨

Three ways to format strings in Python (from newest to oldest):

1. **f-strings** (Python 3.6+) — **recommended** ✅
2. `.format()` method
3. `%` operator (old style)

````python
name = "Priya"
age = 28
gpa = 3.856

# === f-strings (best!) ===
print(f"Name: {name}, Age: {age}")
print(f"Next year: {age + 1}")          # expressions inside {}
print(f"Name uppercase: {name.upper()}")  # method calls too!
````

````python
# === Format specifiers (the real power!) ===
price = 1234.5678
name = "Priya"

# Decimal places
print(f"Price: {price:.2f}")        # 1234.57 (2 decimal places)
print(f"Price: {price:.0f}")        # 1235 (no decimals, rounded)

# Comma separator for thousands
big_number = 1234567890
print(f"Population: {big_number:,}")  # 1,234,567,890

# Percentage
ratio = 0.8567
print(f"Score: {ratio:.1%}")          # 85.7%

# Width & alignment
print(f"|{'left':<20}|")     # |left                |
print(f"|{'center':^20}|")   # |       center       |
print(f"|{'right':>20}|")    # |               right|

# Padding with characters
print(f"|{'hello':*^20}|")   # |*******hello********|
print(f"|{42:0>10}|")        # |0000000042|
````

````python
# === .format() method ===
print("Name: {}, Age: {}".format("Priya", 28))
print("Name: {name}, Age: {age}".format(name="Priya", age=28))
print("Price: {:.2f}".format(99.5))    # 99.50

# Same format specifiers work with .format()
print("{:>20}".format("right aligned"))
````

## 2.6 Common String Patterns 🔧

Let's learn some real-world string tasks you'll encounter often.

````python
# === Pattern 1: Validate input ===
user_input = "12345"
if user_input.isdigit():
    print(f"'{user_input}' is a valid number")
else:
    print(f"'{user_input}' is NOT a valid number")

# === Pattern 2: Extract parts of a string ===
email = "priya.sharma@gmail.com"
username = email.split("@")[0]     # everything before @
domain = email.split("@")[1]       # everything after @
print(f"Username: {username}, Domain: {domain}")

# === Pattern 3: Check prefix/suffix ===
filename = "report_2024.pdf"
if filename.endswith(".pdf"):
    print("This is a PDF file")
if filename.startswith("report"):
    print("This is a report")
````

````python
# === Pattern 4: Build strings from parts ===
words = ["Python", "is", "awesome"]
sentence = " ".join(words)
print(sentence)   # Python is awesome

# Build a path
parts = ["home", "user", "documents", "file.txt"]
path = "/".join(parts)
print(path)       # home/user/documents/file.txt

# === Pattern 5: Pad strings ===
order_id = "42"
padded = order_id.zfill(8)     # zero-fill to 8 digits
print(f"Order: {padded}")     # Order: 00000042

title = "Chapter 1"
print(title.center(30, "─"))   # ──────────Chapter 1──────────
print(title.ljust(30, "."))    # Chapter 1.....................
print(title.rjust(30, "."))    # .....................Chapter 1
````
