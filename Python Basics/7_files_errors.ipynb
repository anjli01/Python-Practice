## 7.5 Error Handling 🛡️

```python
try:
    risky_code()
except SomeError as e:
    handle_error()
else:
    runs_if_no_error()
finally:
    always_runs()
```

````python
# Common exceptions

# ZeroDivisionError
try:
    result = 10 / 0
except ZeroDivisionError:
    print("❌ Can't divide by zero!")

# ValueError
try:
    num = int("hello")
except ValueError as e:
    print(f"❌ Invalid number: {e}")

# FileNotFoundError
try:
    with open("nonexistent.txt") as f:
        content = f.read()
except FileNotFoundError:
    print("❌ File not found!")

# KeyError
try:
    d = {"name": "Priya"}
    print(d["age"])
except KeyError as e:
    print(f"❌ Key not found: {e}")

# IndexError
try:
    lst = [1, 2, 3]
    print(lst[10])
except IndexError:
    print("❌ Index out of range!")
````

````python
# try/except/else/finally

def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("  ❌ Division by zero!")
        return None
    except TypeError as e:
        print(f"  ❌ Type error: {e}")
        return None
    else:
        print(f"  ✅ Result: {result}")   # only runs if NO exception
        return result
    finally:
        print("  📋 Division attempt complete")  # ALWAYS runs

safe_divide(10, 3)
print()
safe_divide(10, 0)
print()
safe_divide("10", 2)
````

### ✏️ Exercise 9 — Handle Division by Zero ➗

````python
# TODO: Write a safe calculator that handles errors

def safe_calc(a, b, operation):
    """Perform a calculation with error handling."""
    try:
        if operation == "+":
            return a + b
        elif operation == "-":
            return a - b
        elif operation == "*":
            return a * b
        elif operation == "/":
            return a / b
        else:
            raise ValueError(f"Unknown operation: {operation}")
    except ZeroDivisionError:
        return "Error: Division by zero!"
    except TypeError:
        return "Error: Invalid types!"
    except ValueError as e:
        return f"Error: {e}"

# Test cases
print(safe_calc(10, 3, "+"))      # 13
print(safe_calc(10, 0, "/"))      # Error: Division by zero!
print(safe_calc("10", 3, "+"))    # 10333 (string concatenation — tricky!)
print(safe_calc(10, 3, "%"))      # Error: Unknown operation: %
````

### ✏️ Exercise 10 — Handle File Not Found 📂

````python
# TODO: Write a function that safely reads a file

def read_file_safe(filepath):
    """Read a file safely, returning content or error message."""
    try:
        with open(filepath, "r") as f:
            content = f.read()
        return True, content
    except FileNotFoundError:
        return False, f"File '{filepath}' not found!"
    except PermissionError:
        return False, f"No permission to read '{filepath}'!"
    except Exception as e:
        return False, f"Unexpected error: {e}"

# Test with existing and non-existing files
files = ["data/sample_text.txt", "data/nonexistent.txt", "data/config.json"]

for filepath in files:
    success, result = read_file_safe(filepath)
    if success:
        print(f"✅ {filepath}: {len(result)} chars")
    else:
        print(f"❌ {filepath}: {result}")
````

### ✏️ Exercise 11 — Custom Input Validator 🛡️

````python
# TODO: Write a robust input function that keeps asking until valid

def get_valid_number(prompt, min_val=None, max_val=None, num_type=float):
    """Get a valid number from user with range checking."""
    while True:
        try:
            value = num_type(input(prompt))
            
            # Check range
            if min_val is not None and value < min_val:
                print(f"  ⚠️ Must be at least {min_val}")
                continue
            if max_val is not None and value > max_val:
                print(f"  ⚠️ Must be at most {max_val}")
                continue
            
            return value
        except ValueError:
            print(f"  ⚠️ Please enter a valid number!")

# Test it
age = get_valid_number("Enter age (1-120): ", min_val=1, max_val=120, num_type=int)
print(f"Age: {age}")

temp = get_valid_number("Enter temperature (-50 to 50): ", min_val=-50, max_val=50)
print(f"Temperature: {temp}°C")
````

### ✏️ Exercise 13 — Build a File-Based To-Do List ✅

````python
import json
from datetime import datetime

TODO_FILE = "data/todos.json"

def load_todos():
    """Load todos from JSON file."""
    try:
        with open(TODO_FILE, "r") as f:
            return json.load(f)
    except (FileNotFoundError, json.JSONDecodeError):
        return []

def save_todos(todos):
    """Save todos to JSON file."""
    with open(TODO_FILE, "w") as f:
        json.dump(todos, f, indent=2)

def show_todos(todos):
    if not todos:
        print("  📭 No tasks yet!")
        return
    for i, todo in enumerate(todos, 1):
        icon = "✅" if todo["done"] else "⬜"
        print(f"  {i}. {icon} {todo['task']} (added: {todo['added']})")

def add_todo(todos):
    task = input("  New task: ").strip()
    if task:
        todos.append({
            "task": task,
            "done": False,
            "added": datetime.now().strftime("%Y-%m-%d %H:%M")
        })
        save_todos(todos)
        print(f"  ✅ Added: {task}")

def complete_todo(todos):
    show_todos(todos)
    try:
        num = int(input("  Complete task #: "))
        if 1 <= num <= len(todos):
            todos[num - 1]["done"] = True
            save_todos(todos)
            print(f"  ✅ Completed: {todos[num-1]['task']}")
        else:
            print("  ⚠️ Invalid number!")
    except ValueError:
        print("  ⚠️ Enter a number!")

# Main loop
print("📋 TO-DO LIST")
todos = load_todos()

while True:
    print("\n1. View  2. Add  3. Complete  4. Quit")
    choice = input("Choose: ").strip()
    if choice == "1": show_todos(todos)
    elif choice == "2": add_todo(todos)
    elif choice == "3": complete_todo(todos)
    elif choice == "4":
        print("👋 Bye!")
        break
````

## 🎯 Exercise 14 — Capstone Project 1: Expense Tracker 💰

A complete CRUD application with file persistence. Uses: functions, dicts, lists, file I/O, CSV, error handling, loops, and formatting.

````python
import csv
from datetime import datetime

# =============================================
#  🎯 CAPSTONE 1: Expense Tracker
# =============================================

EXPENSE_FILE = "data/expenses.csv"

def load_expenses():
    """Load expenses from CSV."""
    expenses = []
    try:
        with open(EXPENSE_FILE, "r") as f:
            reader = csv.DictReader(f)
            for row in reader:
                row["amount"] = float(row["amount"])
                expenses.append(row)
    except FileNotFoundError:
        pass
    return expenses

def save_expenses(expenses):
    """Save all expenses to CSV."""
    with open(EXPENSE_FILE, "w", newline="") as f:
        fieldnames = ["date", "category", "description", "amount"]
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        for exp in expenses:
            writer.writerow(exp)

def add_expense(expenses):
    """Add a new expense."""
    categories = ["Food", "Transport", "Shopping", "Bills", "Entertainment", "Other"]
    print("  Categories:", ", ".join(f"{i}.{c}" for i, c in enumerate(categories, 1)))
    
    try:
        cat_num = int(input("  Category (1-6): "))
        category = categories[cat_num - 1]
    except (ValueError, IndexError):
        category = "Other"
    
    desc = input("  Description: ").strip()
    try:
        amount = float(input("  Amount (₹): "))
    except ValueError:
        print("  ⚠️ Invalid amount!")
        return
    
    expense = {
        "date": datetime.now().strftime("%Y-%m-%d"),
        "category": category,
        "description": desc,
        "amount": amount
    }
    expenses.append(expense)
    save_expenses(expenses)
    print(f"  ✅ Added: {desc} - ₹{amount:,.2f} ({category})")

def view_expenses(expenses):
    """Display all expenses."""
    if not expenses:
        print("  📭 No expenses recorded!")
        return
    print(f"  {'Date':<12} {'Category':<15} {'Description':<20} {'Amount':>10}")
    print("  " + "-" * 60)
    total = 0
    for exp in expenses:
        print(f"  {exp['date']:<12} {exp['category']:<15} {exp['description']:<20} ₹{exp['amount']:>8,.2f}")
        total += exp['amount']
    print("  " + "-" * 60)
    print(f"  {'TOTAL':<48} ₹{total:>8,.2f}")

def view_summary(expenses):
    """Show spending by category."""
    if not expenses:
        print("  📭 No data!")
        return
    by_cat = {}
    for exp in expenses:
        cat = exp["category"]
        by_cat[cat] = by_cat.get(cat, 0) + exp["amount"]
    
    total = sum(by_cat.values())
    print("\n  📊 SPENDING BY CATEGORY:")
    for cat, amount in sorted(by_cat.items(), key=lambda x: -x[1]):
        pct = amount / total * 100
        bar = "█" * int(pct / 2)
        print(f"  {cat:<15} ₹{amount:>8,.0f} ({pct:>5.1f}%) {bar}")
    print(f"\n  💰 Total spent: ₹{total:,.2f}")

# --- Main loop ---
print("💰 EXPENSE TRACKER")
expenses = load_expenses()

while True:
    print("\n1. Add  2. View All  3. Summary  4. Quit")
    choice = input("Choose: ").strip()
    if choice == "1": add_expense(expenses)
    elif choice == "2": view_expenses(expenses)
    elif choice == "3": view_summary(expenses)
    elif choice == "4":
        print("👋 Goodbye!")
        break
````

## 🎯 Exercise 15 — Capstone Project 2: Word Frequency Analyzer 📊

Read a text file, analyze its content, and generate a formatted report. Uses: file I/O, strings, dicts, sorting, functions, and formatting.

````python
import json
from datetime import datetime

# =============================================
#  🎯 CAPSTONE 2: Word Frequency Analyzer
# =============================================

def read_text(filepath):
    """Read text from a file."""
    try:
        with open(filepath, "r") as f:
            return f.read()
    except FileNotFoundError:
        print(f"❌ File not found: {filepath}")
        return None

def count_words(text):
    """Count word frequencies (case-insensitive, stripped of punctuation)."""
    import string
    # Remove punctuation and lowercase
    cleaned = text.lower()
    for char in string.punctuation:
        cleaned = cleaned.replace(char, "")
    
    words = cleaned.split()
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
    return freq

def count_chars(text):
    """Count character frequencies (letters only)."""
    freq = {}
    for char in text.lower():
        if char.isalpha():
            freq[char] = freq.get(char, 0) + 1
    return freq

def generate_report(filepath):
    """Generate a complete text analysis report."""
    text = read_text(filepath)
    if not text:
        return
    
    # Basic stats
    lines = [l for l in text.split("\n") if l.strip()]
    all_words = text.split()
    word_freq = count_words(text)
    char_freq = count_chars(text)
    
    # Top words
    top_words = sorted(word_freq.items(), key=lambda x: -x[1])[:15]
    top_chars = sorted(char_freq.items(), key=lambda x: -x[1])[:10]
    
    # Unique words
    unique_count = len(word_freq)
    
    # Average word length
    avg_len = sum(len(w) for w in all_words) / len(all_words)
    
    # Print report
    print("═" * 55)
    print(f"{'📊 TEXT ANALYSIS REPORT':^55}")
    print(f"{'File: ' + filepath:^55}")
    print("═" * 55)
    
    print(f"\n📏 BASIC STATISTICS")
    print(f"  Lines:            {len(lines):>8}")
    print(f"  Total words:      {len(all_words):>8}")
    print(f"  Unique words:     {unique_count:>8}")
    print(f"  Characters:       {len(text):>8}")
    print(f"  Avg word length:  {avg_len:>8.1f}")
    
    print(f"\n📈 TOP 15 WORDS")
    max_count = top_words[0][1] if top_words else 1
    for word, count in top_words:
        bar_len = int(count / max_count * 20)
        bar = "█" * bar_len
        print(f"  {word:<15} {count:>3} {bar}")
    
    print(f"\n🔤 TOP 10 CHARACTERS")
    total_chars = sum(char_freq.values())
    for char, count in top_chars:
        pct = count / total_chars * 100
        bar = "█" * int(pct)
        print(f"  '{char}' {count:>4} ({pct:>5.1f}%) {bar}")
    
    print("\n" + "═" * 55)
    
    # Save report as JSON
    report_data = {
        "file": filepath,
        "analyzed_at": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "stats": {
            "lines": len(lines),
            "words": len(all_words),
            "unique_words": unique_count,
            "characters": len(text),
            "avg_word_length": round(avg_len, 2)
        },
        "top_words": dict(top_words),
        "top_characters": dict(top_chars),
    }
    
    with open("data/analysis_report.json", "w") as f:
        json.dump(report_data, f, indent=2)
    print("\n✅ Report saved to data/analysis_report.json")

# Run the analyzer!
generate_report("data/sample_text.txt")
````
