## 1.2 `print()` and Comments 🖨️

`print()` is the most basic command — it displays output on screen.

**Comments** are notes for humans. Python ignores them completely.
- Single line: `# this is a comment`
- Multi-line: use triple quotes `"""like this"""`

````python
# === Basic print ===
print("I love coffee!")          # A simple string
print(42)                         # A number
print(3.14)                       # A decimal
print(True)                       # A boolean
````

````python
# === Multiple arguments ===
print("Name:", "Priya", "| Age:", 28)

# By default, print() separates items with a space
# You can change that with 'sep':
print("2025", "10", "18", sep="-")    # 2025-10-18
print("apple", "banana", "cherry", sep=" 🍎 ")  # custom separator
````

````python
# === The 'end' parameter ===
# By default, print() ends with a newline (\n)
# You can change that:
print("Loading", end="")
print("...", end="")
print("...", end="")
print(" Done! ✅")
````

````python
# === Multi-line strings (triple quotes) ===
poem = """Roses are red,
Violets are blue,
Python is awesome,
And so are you! 💐"""

print(poem)
````

````python
# === Comments ===

# This is a single-line comment — Python skips it

"""
This is a multi-line comment (technically a multi-line string).
Useful for longer notes or temporarily disabling code.
"""

print("Comments don't appear in output!")  # This part is also a comment
````

## 1.3 Variables & Naming Rules 📦

A **variable** stores a value so you can use it later. Think of it as a **labelled box**.

```python
name = "Priya"   # name is the label, "Priya" is the value
```

### Naming Rules

| ✅ Valid | ❌ Invalid | Why? |
|---------|-----------|------|
| `my_name` | `my name` | No spaces |
| `age2` | `2age` | Can't start with a number |
| `_private` | `my-var` | No hyphens |
| `firstName` | `class` | `class` is a reserved word |

**Convention:** Use `snake_case` (lowercase with underscores) → `total_price`, `user_name`

````python
# Creating variables
city = "Mumbai"
population = 20_000_000    # underscores make big numbers readable!
area_sq_km = 603.4
is_capital = False

print(f"{city} has a population of {population}")
print(f"Area: {area_sq_km} sq km")
print(f"Is it the capital? {is_capital}")
````

````python
# Variables can be updated (reassigned)
score = 0
print(f"Initial score: {score}")

score = 10
print(f"After round 1: {score}")

score = score + 5       # old value + 5
print(f"After round 2: {score}")

score += 3              # shorthand for score = score + 3
print(f"After bonus: {score}")
````

````python
# Multiple assignment in one line
x, y, z = 10, 20, 30
print(x, y, z)

# Same value to multiple variables
a = b = c = 0
print(a, b, c)
````

## 1.4 Data Types 🏷️

Every value in Python has a **type**. The four basic types are:

| Type | Python Name | Examples | Use For |
|------|-------------|---------|--------|
| Integer | `int` | `42`, `-5`, `0` | Whole numbers, counting |
| Float | `float` | `3.14`, `-0.5`, `2.0` | Decimals, measurements |
| String | `str` | `"hello"`, `'world'` | Text, names, messages |
| Boolean | `bool` | `True`, `False` | Yes/No, On/Off |

````python
# type() tells you what kind of data a value is
print(type(42))          # <class 'int'>
print(type(3.14))        # <class 'float'>
print(type("hello"))     # <class 'str'>
print(type(True))        # <class 'bool'>

# Variables also have types
price = 99.99
print(type(price))       # <class 'float'>
````

````python
# isinstance() — check if a value is a certain type
age = 25
print(isinstance(age, int))      # True
print(isinstance(age, str))      # False
print(isinstance(age, float))    # False

# Booleans are secretly integers!
print(isinstance(True, int))     # True (surprise!)
print(True + True)               # 2
print(False + 10)                # 10
````

## 1.5 Type Conversion 🔄

You can convert between types:

| Function | Converts to | Example |
|----------|------------|--------|
| `int()` | Integer | `int("42")` → `42` |
| `float()` | Float | `float("3.14")` → `3.14` |
| `str()` | String | `str(42)` → `"42"` |
| `bool()` | Boolean | `bool(1)` → `True` |

````python
# === String ↔ Number conversions ===
text_number = "42"
real_number = int(text_number)
print(real_number + 8)           # 50 (math works now!)

price = 99.99
price_text = str(price)
print("Price: ₹" + price_text)   # Price: ₹99.99

# int() truncates (doesn't round!)
print(int(3.9))    # 3 (not 4!)
print(int(-3.9))   # -3
````

````python
# === bool() — what's truthy and falsy? ===
# These are FALSY (convert to False):
print(bool(0))        # False
print(bool(0.0))      # False
print(bool(""))       # False (empty string)
print(bool(None))     # False
print(bool([]))       # False (empty list)

print("---")

# Everything else is TRUTHY (converts to True):
print(bool(1))        # True
print(bool(-5))       # True (any non-zero number)
print(bool("hello"))  # True (any non-empty string)
print(bool(" "))      # True (even a space!)
````

## 1.6 Arithmetic Operators ➕➖✖️➗

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `7 + 3` | `10` |
| `-` | Subtraction | `7 - 3` | `4` |
| `*` | Multiplication | `7 * 3` | `21` |
| `/` | Division | `7 / 3` | `2.333...` |
| `//` | Floor division | `7 // 3` | `2` |
| `%` | Modulo (remainder) | `7 % 3` | `1` |
| `**` | Exponentiation | `7 ** 3` | `343` |

### Operator Precedence (BODMAS / PEMDAS)
`**` → `* / // %` → `+ -`  (left to right within same level)

````python
# Basic arithmetic
print(10 + 3)     # 13
print(10 - 3)     # 7
print(10 * 3)     # 30
print(10 / 3)     # 3.3333... (always gives float!)
print(10 // 3)    # 3 (rounds DOWN)
print(10 % 3)     # 1 (remainder)
print(10 ** 3)    # 1000
````

````python
# Operator precedence
print(2 + 3 * 4)        # 14 (not 20!) — multiplication first
print((2 + 3) * 4)      # 20 — parentheses override precedence
print(2 ** 3 ** 2)       # 512 — exponentiation is RIGHT to LEFT: 3**2=9, 2**9=512
print(10 - 2 - 3)       # 5 — subtraction is LEFT to RIGHT
````

````python
# Augmented assignment operators (shortcuts)
score = 100
score += 10     # same as: score = score + 10
print(score)    # 110

score -= 25     # same as: score = score - 25
print(score)    # 85

score *= 2      # same as: score = score * 2
print(score)    # 170

score //= 3     # same as: score = score // 3
print(score)    # 56
````

## 1.7 Comparison & Logical Operators ⚖️🔗

### Comparison Operators (return `True` or `False`)

| Operator | Meaning | Example |
|----------|---------|--------|
| `==` | Equal to | `5 == 5` → `True` |
| `!=` | Not equal to | `5 != 3` → `True` |
| `>` | Greater than | `5 > 3` → `True` |
| `<` | Less than | `5 < 3` → `False` |
| `>=` | Greater or equal | `5 >= 5` → `True` |
| `<=` | Less or equal | `5 <= 3` → `False` |

### Logical Operators (combine conditions)

| Operator | Meaning | Example |
|----------|---------|--------|
| `and` | Both must be True | `True and False` → `False` |
| `or` | At least one True | `True or False` → `True` |
| `not` | Flip True/False | `not True` → `False` |

````python
# Comparisons
print(10 > 5)       # True
print(10 == 10)     # True (== checks equality, = assigns)
print(10 != 10)     # False
print("apple" < "banana")   # True (alphabetical comparison)

# Python allows chained comparisons!
age = 25
print(18 <= age <= 65)      # True — is age between 18 and 65?
````

````python
# Logical operators
age = 25
has_id = True

print(age >= 18 and has_id)      # True (both conditions are true)
print(age >= 18 or has_id)       # True (at least one is true)
print(not has_id)                 # False (flips True to False)

# Common pattern: check if a value is in a range
temperature = 22
is_comfortable = 20 <= temperature <= 26
print(f"Is {temperature}°C comfortable? {is_comfortable}")  # True
````

### Exercise 7 — Tricky Type Conversions
```
int("42")      → 42 ✅
int("42.5")    → ERROR! ❌ (can't parse decimal string)
int("hello")   → ERROR! ❌ (not a number at all)
float("42")    → 42.0 ✅
str(42)        → "42" ✅
bool("")       → False (empty string is falsy)
bool("False")  → True! (non-empty string, even if it says "False")
int(True)      → 1
int(False)     → 0
```

### Exercise 12 — Comparison Chain Predictions
```
x == z       → True  (10 == 10)
x != y       → True  (10 != 20)
x > y        → False (10 > 20)
x >= z       → True  (10 >= 10)
1 < x < 15   → True  (1 < 10 < 15)
x == y == z   → False (10 != 20)
"cat" < "dog" → True  (alphabetical)
"Zebra" < "apple" → True (uppercase Z < lowercase a in ASCII)
```

### Exercise 13 — Logical Operator Truth Table
```
AND: T&T=T, T&F=F, F&T=F, F&F=F
OR:  T|T=T, T|F=T, F|T=T, F|F=F
NOT: !T=F, !F=T

not True or False         → False (not has higher precedence)
not (True or False)       → False
True and not False        → True
(True or False) and (not False) → True
```
