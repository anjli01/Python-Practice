## 5.1 Dict Creation & Access 📖

A **dictionary** maps **keys** to **values** — like a real dictionary maps words to definitions.

```python
my_dict = {"key1": value1, "key2": value2}
```

````python
# Creating dictionaries
person = {
    "name": "Priya",
    "age": 28,
    "city": "Mumbai",
    "is_student": False
}

print(person)
print(f"Type: {type(person)}")
print(f"Length: {len(person)} key-value pairs")

# Accessing values
print(f"Name: {person['name']}")
print(f"Age: {person['age']}")

# ⚠️ KeyError if key doesn't exist:
# print(person["email"])   # ❌ KeyError!
````

````python
# .get() — safe access (returns None or default if key missing)
print(person.get("name"))           # Priya
print(person.get("email"))          # None (no error!)
print(person.get("email", "N/A"))   # N/A (custom default)

# Check if key exists
print("name" in person)     # True
print("email" in person)    # False

# Different ways to create dicts
d1 = {"a": 1, "b": 2}
d2 = dict(a=1, b=2)              # using dict()
d3 = dict([("a", 1), ("b", 2)])  # from list of tuples
print(d1 == d2 == d3)             # True
````

## 5.2 Dict Modification ✏️

Dictionaries are **mutable** — you can add, update, and delete key-value pairs.

````python
user = {"name": "Priya", "age": 28}

# Add new key
user["email"] = "priya@mail.com"

# Update existing key
user["age"] = 29

# Update multiple keys at once
user.update({"city": "Mumbai", "age": 30})

# Delete a key
del user["email"]

# Pop — remove and return value
city = user.pop("city")
print(f"Removed: {city}")

# setdefault — only set if key doesn't exist
user.setdefault("country", "India")  # adds it
user.setdefault("name", "Unknown")   # does nothing (name already exists)

print(user)
````

## 5.3 Dict Iteration 🔄

Three ways to loop through a dictionary.

````python
scores = {"Math": 92, "Science": 88, "English": 95, "History": 78}

# Loop through keys (default)
print("=== Keys ===")
for key in scores:
    print(f"  {key}")

# Loop through values
print("\n=== Values ===")
for value in scores.values():
    print(f"  {value}")

# Loop through key-value pairs ✅ (most useful)
print("\n=== Items ===")
for subject, score in scores.items():
    print(f"  {subject}: {score}")

# You can also get them as lists
print(f"\nKeys: {list(scores.keys())}")
print(f"Values: {list(scores.values())}")
````

## 5.5 Sets 🔵

A **set** is an **unordered** collection of **unique** items. No duplicates!

| Feature | List | Set |
|---------|------|-----|
| Ordered? | ✅ Yes | ❌ No |
| Duplicates? | ✅ Allowed | ❌ No duplicates |
| Mutable? | ✅ Yes | ✅ Yes |
| Syntax | `[1, 2, 3]` | `{1, 2, 3}` |

````python
# Creating sets
fruits = {"apple", "banana", "cherry", "apple"}  # duplicate removed!
print(fruits)        # {'apple', 'banana', 'cherry'}
print(len(fruits))   # 3

# From a list (removes duplicates!)
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
unique = set(numbers)
print(f"Unique: {unique}")  # {1, 2, 3, 4}

# Empty set (NOT {} — that's an empty dict!)
empty_set = set()
empty_dict = {}
print(f"Set type: {type(empty_set)}, Dict type: {type(empty_dict)}")

# Add and remove
colors = {"red", "green", "blue"}
colors.add("yellow")
colors.discard("green")     # remove (no error if missing)
# colors.remove("purple")   # ❌ would raise KeyError
colors.discard("purple")    # ✅ no error
print(colors)
````

````python
# === Set Operations (like Venn diagrams!) ===
python_devs = {"Anita", "Rahul", "Sneha", "Priya"}
js_devs = {"Rahul", "Karan", "Sneha", "Vishal"}

# Union — everyone (in either or both)
all_devs = python_devs | js_devs
print(f"All devs: {all_devs}")

# Intersection — in BOTH
both = python_devs & js_devs
print(f"Know both: {both}")

# Difference — in Python but NOT JS
only_python = python_devs - js_devs
print(f"Python only: {only_python}")

# Symmetric difference — in one but NOT both
exclusive = python_devs ^ js_devs
print(f"Exclusive: {exclusive}")

# Subset/superset checks
print(f"Is {{1,2}} a subset of {{1,2,3}}? { {1,2} <= {1,2,3} }")
````

## 5.6 Combining Data Structures 🏗️

Real-world data often requires **lists of dicts**, **dicts of lists**, or other combinations.

````python
# List of dicts — like rows in a spreadsheet
employees = [
    {"name": "Anita", "dept": "Engineering", "salary": 95000},
    {"name": "Rahul", "dept": "Marketing", "salary": 75000},
    {"name": "Sneha", "dept": "Engineering", "salary": 105000},
    {"name": "Vishal", "dept": "Marketing", "salary": 70000},
    {"name": "Priya", "dept": "Engineering", "salary": 92000},
]

# Access specific fields
for emp in employees:
    print(f"{emp['name']:<10} {emp['dept']:<15} ₹{emp['salary']:,}")

# Get all names
names = [emp["name"] for emp in employees]
print(f"\nAll names: {names}")

# Average salary
avg_sal = sum(emp["salary"] for emp in employees) / len(employees)
print(f"Avg salary: ₹{avg_sal:,.0f}")
````

## 🎯 Exercise 15 — Mini-Project: Contact Book 📱

````python
# =============================================
#    🎯 MINI-PROJECT: Contact Book
# =============================================

contacts = {
    "Anita": {"phone": "+91-98765-11111", "email": "anita@mail.com", "city": "Mumbai"},
    "Rahul": {"phone": "+91-98765-22222", "email": "rahul@mail.com", "city": "Delhi"},
}

def show_contacts():
    if not contacts:
        print("  📭 No contacts yet!")
        return
    print(f"  {'Name':<12} {'Phone':<18} {'Email':<22} {'City'}")
    print("  " + "-" * 62)
    for name, info in sorted(contacts.items()):
        print(f"  {name:<12} {info['phone']:<18} {info['email']:<22} {info['city']}")

def add_contact():
    name = input("  Name: ").strip().title()
    if name in contacts:
        print(f"  ⚠️ {name} already exists!")
        return
    phone = input("  Phone: ").strip()
    email = input("  Email: ").strip()
    city = input("  City: ").strip().title()
    contacts[name] = {"phone": phone, "email": email, "city": city}
    print(f"  ✅ {name} added!")

def search_contact():
    query = input("  Search name: ").strip().lower()
    results = {n: i for n, i in contacts.items() if query in n.lower()}
    if results:
        for name, info in results.items():
            print(f"  📇 {name}: {info['phone']} | {info['email']} | {info['city']}")
    else:
        print("  ❌ No matches found.")

def delete_contact():
    name = input("  Name to delete: ").strip().title()
    if name in contacts:
        del contacts[name]
        print(f"  🗑️ {name} deleted.")
    else:
        print(f"  ❌ {name} not found.")

# Main loop
print("📱 CONTACT BOOK")
while True:
    print("\n1. View all  2. Add  3. Search  4. Delete  5. Quit")
    choice = input("Choose: ").strip()
    
    if choice == "1":
        show_contacts()
    elif choice == "2":
        add_contact()
    elif choice == "3":
        search_contact()
    elif choice == "4":
        delete_contact()
    elif choice == "5":
        print("👋 Goodbye!")
        break
    else:
        print("  ⚠️ Invalid choice!")
````
