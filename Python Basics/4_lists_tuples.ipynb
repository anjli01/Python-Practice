## 4.1 List Creation & Access 📋

A **list** is an ordered, mutable collection of items. It can hold any type of data.

````python
# Creating lists
fruits = ["Mango", "Apple", "Banana", "Grapes"]
numbers = [10, 20, 30, 40, 50]
mixed = ["hello", 42, 3.14, True, None]    # different types!
empty = []                                    # empty list

print(f"Fruits: {fruits}")
print(f"Length: {len(fruits)}")
print(f"Type: {type(fruits)}")
````

````python
# Indexing — same as strings!
colors = ["Red", "Green", "Blue", "Yellow", "Purple"]

print(colors[0])      # Red (first)
print(colors[-1])     # Purple (last)
print(colors[2])      # Blue

# Slicing
print(colors[1:4])    # ['Green', 'Blue', 'Yellow']
print(colors[:3])     # ['Red', 'Green', 'Blue']
print(colors[::2])    # ['Red', 'Blue', 'Purple'] (every 2nd)
print(colors[::-1])   # reversed list
````

````python
# Nested lists (list of lists)
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[0])       # [1, 2, 3] (first row)
print(matrix[1][2])    # 6 (row 1, column 2)
print(matrix[2][0])    # 7 (row 2, column 0)

# Print as a grid
for row in matrix:
    for val in row:
        print(f"{val:>3}", end="")
    print()
````

## 4.2 List Modification 🔨

Lists are **mutable** — you can change them in place.

| Method | Action | Example |
|--------|--------|--------|
| `.append(x)` | Add to end | `[1,2].append(3)` → `[1,2,3]` |
| `.insert(i, x)` | Insert at position | `[1,3].insert(1, 2)` → `[1,2,3]` |
| `.extend(list)` | Add multiple items | `[1].extend([2,3])` → `[1,2,3]` |
| `.remove(x)` | Remove first occurrence | `[1,2,2].remove(2)` → `[1,2]` |
| `.pop(i)` | Remove & return item at index | `[1,2,3].pop(1)` → returns `2` |
| `.pop()` | Remove & return last item | `[1,2,3].pop()` → returns `3` |
| `del list[i]` | Delete item at index | `del [1,2,3][0]` → `[2,3]` |
| `.clear()` | Remove all items | `[1,2,3].clear()` → `[]` |

````python
# Demonstration of all modification methods
tasks = ["Buy groceries", "Clean house"]
print(f"Start: {tasks}")

tasks.append("Cook dinner")
print(f"After append: {tasks}")

tasks.insert(1, "Do laundry")
print(f"After insert: {tasks}")

tasks.extend(["Read book", "Exercise"])
print(f"After extend: {tasks}")

tasks.remove("Clean house")
print(f"After remove: {tasks}")

done = tasks.pop(0)     # removes and returns first item
print(f"Completed: {done}")
print(f"After pop: {tasks}")
````

## 4.3 List Operations 🔧

Sorting, reversing, copying, and more.

````python
scores = [85, 92, 78, 95, 88, 72, 90]

# sorted() — returns a NEW sorted list (original unchanged)
print(f"Sorted: {sorted(scores)}")
print(f"Original: {scores}")          # unchanged!

# .sort() — sorts in place (modifies original)
scores.sort()
print(f"After .sort(): {scores}")      # changed!

# Reverse sort
scores.sort(reverse=True)
print(f"Descending: {scores}")

# .reverse() — reverses in place
names = ["Zara", "Anita", "Priya"]
names.reverse()
print(f"Reversed: {names}")

# Useful functions
nums = [23, 45, 12, 67, 34]
print(f"Sum: {sum(nums)}, Min: {min(nums)}, Max: {max(nums)}")
print(f"Index of 67: {nums.index(67)}")
print(f"Count of 12: {nums.count(12)}")
````

````python
# IMPORTANT: Copying lists
# ⚠️ Assignment does NOT copy — both names point to the SAME list!
a = [1, 2, 3]
b = a           # b is NOT a copy — it's the same list!
b.append(4)
print(f"a: {a}")  # [1, 2, 3, 4] — a was also changed! 😱
print(f"b: {b}")  # [1, 2, 3, 4]

# To make a TRUE copy:
c = [1, 2, 3]
d = c.copy()     # method 1: .copy()
e = c[:]         # method 2: slice
f = list(c)      # method 3: list()

d.append(99)
print(f"c: {c}")  # [1, 2, 3] — unchanged! ✅
print(f"d: {d}")  # [1, 2, 3, 99]
````

## 4.4 List Comprehensions ⚡

A **list comprehension** is a compact way to create lists. It replaces the **loop + append** pattern.

```python
# Loop pattern:
result = []
for item in iterable:
    result.append(expression)

# Comprehension equivalent:
result = [expression for item in iterable]
```

````python
# Basic list comprehension
squares = [x ** 2 for x in range(1, 11)]
print(f"Squares: {squares}")

# With condition (filter)
evens = [x for x in range(1, 21) if x % 2 == 0]
print(f"Evens: {evens}")

# Transform + filter
names = ["alice", "bob", "charlie", "diana"]
long_names = [name.title() for name in names if len(name) > 3]
print(f"Long names (titled): {long_names}")

# if-else in comprehension (note: goes BEFORE 'for')
labels = ["even" if x % 2 == 0 else "odd" for x in range(1, 6)]
print(f"Labels: {labels}")
````

````python
# Nested comprehension
# Flatten a matrix
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [num for row in matrix for num in row]
print(f"Flat: {flat}")

# Create a multiplication table
table = [[i * j for j in range(1, 6)] for i in range(1, 6)]
for row in table:
    print(row)
````

## 4.5 Common List Patterns 🧩

````python
# === zip() — iterate over multiple lists in parallel ===
names = ["Anita", "Rahul", "Sneha"]
ages =  [25, 30, 28]
cities = ["Mumbai", "Delhi", "Chennai"]

for name, age, city in zip(names, ages, cities):
    print(f"{name} is {age} from {city}")

# Create pairs
pairs = list(zip(names, ages))
print(f"\nPairs: {pairs}")
````

````python
# === any() and all() ===
scores = [85, 92, 78, 95, 60]

print(all(s >= 60 for s in scores))   # True (ALL passed?)
print(all(s >= 80 for s in scores))   # False (not all above 80)
print(any(s == 100 for s in scores))  # False (any perfect score?)
print(any(s >= 90 for s in scores))   # True (any above 90?)
````

## 4.6 Tuples 📌

A **tuple** is like a list but **immutable** (cannot be changed after creation).

| Feature | List | Tuple |
|---------|------|-------|
| Syntax | `[1, 2, 3]` | `(1, 2, 3)` |
| Mutable? | ✅ Yes | ❌ No |
| Use case | Dynamic collections | Fixed data, dictionary keys, function returns |

````python
# Creating tuples
point = (3, 4)
rgb = (255, 128, 0)
single = (42,)         # single-element tuple NEEDS a comma!
not_tuple = (42)       # this is just the number 42 in parentheses

print(f"Point: {point}, type: {type(point)}")
print(f"Single: {single}, type: {type(single)}")
print(f"Not a tuple: {not_tuple}, type: {type(not_tuple)}")

# Indexing works
print(f"x: {point[0]}, y: {point[1]}")

# But modification doesn't!
# point[0] = 5  # ❌ TypeError: 'tuple' object does not support item assignment
````

````python
# === Tuple unpacking ===
# Assign each element to a variable in one line
name, age, city = ("Priya", 28, "Mumbai")
print(f"{name}, {age}, {city}")

# Swap variables (uses tuple unpacking!)
a, b = 10, 20
a, b = b, a
print(f"a={a}, b={b}")   # a=20, b=10

# Star unpacking — collect extras
first, *rest = [1, 2, 3, 4, 5]
print(f"First: {first}, Rest: {rest}")  # 1, [2, 3, 4, 5]

first, *middle, last = [1, 2, 3, 4, 5]
print(f"First: {first}, Middle: {middle}, Last: {last}")  # 1, [2, 3, 4], 5

# Functions often return tuples
def min_max(numbers):
    return min(numbers), max(numbers)

lo, hi = min_max([4, 2, 9, 1, 7])
print(f"Min: {lo}, Max: {hi}")
````

````python
# =============================================
#   🎯 MINI-PROJECT: Student Grades Manager
# =============================================

# Student data: (name, [math, science, english, history])
students = [
    ("Anita",  [88, 92, 78, 85]),
    ("Rahul",  [72, 68, 81, 74]),
    ("Sneha",  [95, 91, 89, 93]),
    ("Vishal", [65, 70, 58, 62]),
    ("Priya",  [90, 85, 92, 88]),
    ("Karan",  [78, 82, 76, 80]),
]

subjects = ["Math", "Science", "English", "History"]

# --- STEP 1: Calculate averages ---
# TODO: Create a list of (name, average) tuples
averages = []
for name, scores in students:
    avg = ___
    averages.append((name, round(avg, 1)))

# --- STEP 2: Assign grades ---
def get_grade(avg):
    if avg >= 90: return "A"
    elif avg >= 80: return "B"
    elif avg >= 70: return "C"
    elif avg >= 60: return "D"
    else: return "F"

# --- STEP 3: Print report card ---
print("═" * 65)
print(f"{'📊 STUDENT REPORT CARD':^65}")
print("═" * 65)
print(f"{'Name':<10}", end="")
for subj in subjects:
    print(f"{subj:>9}", end="")
print(f"{'  Avg':>7}{'Grade':>7}")
print("-" * 65)

for name, scores in students:
    avg = sum(scores) / len(scores)
    grade = get_grade(avg)
    print(f"{name:<10}", end="")
    for score in scores:
        print(f"{score:>9}", end="")
    print(f"{avg:>7.1f}{'  ' + grade:>5}")

# --- STEP 4: Statistics ---
print("\n" + "-" * 65)

# TODO: Find top student (highest average)
top_name, top_avg = max(averages, key=lambda x: x[1])
print(f"🏆 Top student: {top_name} ({top_avg})")

# TODO: Subject-wise averages
print("\n📚 Subject Averages:")
for i, subj in enumerate(subjects):
    subj_scores = [scores[i] for _, scores in students]
    subj_avg = sum(subj_scores) / len(subj_scores)
    print(f"  {subj:<10}: {subj_avg:.1f}")

# TODO: Students who passed all subjects (>= 60 in every subject)
all_pass = [name for name, scores in students if all(s >= 60 for s in scores)]
print(f"\n✅ Passed all subjects: {all_pass}")

# TODO: Students sorted by average (descending)
ranked = sorted(averages, key=lambda x: x[1], reverse=True)
print("\n🏅 Rankings:")
for rank, (name, avg) in enumerate(ranked, 1):
    print(f"  {rank}. {name:<10} {avg}")

print("\n" + "═" * 65)
````
