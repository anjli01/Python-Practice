## 3.1 Conditionals: if / elif / else 🔀

Conditionals let Python **choose** which code to run based on conditions.

```python
if condition:       # checked first
    do_this
elif condition2:    # checked only if 'if' was False
    do_that
else:               # runs if ALL above were False
    do_default
```

⚠️ **Indentation matters!** Use 4 spaces for code inside `if/elif/else`.

````python
# === Basic if/elif/else ===
weather = "rainy"

if weather == "sunny":
    print("☀️ Wear sunglasses!")
elif weather == "rainy":
    print("🌧️ Take an umbrella!")
elif weather == "snowy":
    print("❄️ Wear a warm coat!")
else:
    print("🤷 Check the weather app!")
````

````python
# === Nested if ===
age = 25
has_license = True
has_car = False

if age >= 18:
    print("You're an adult.")
    if has_license:
        print("You have a license.")
        if has_car:
            print("You can drive your own car! 🚗")
        else:
            print("You can drive, but need to borrow a car. 🔑")
    else:
        print("Get a license first! 📋")
else:
    print("You're not old enough to drive.")
````

````python
# === Ternary operator (one-line if/else) ===
age = 20
status = "adult" if age >= 18 else "minor"
print(f"You are a {status}")    # You are a adult

# Useful for quick assignments
score = 85
grade = "Pass" if score >= 60 else "Fail"
print(f"Score: {score} → {grade}")  # Pass

# You can even use it in print
temp = 35
print(f"It's {'hot 🥵' if temp > 30 else 'nice 😊' if temp > 20 else 'cold 🥶'}")
````

## 3.2 `for` Loops 🔄

A `for` loop iterates through a **sequence** (list, string, range, etc.).

````python
# Loop through a list
colors = ["Red", "Green", "Blue"]
for color in colors:
    print(f"I see {color}!")

print("---")

# Loop through a string
for char in "Python":
    print(char, end=" ")  # P y t h o n
print()
````

````python
# === range() ===
# range(stop)           → 0, 1, 2, ..., stop-1
# range(start, stop)    → start, start+1, ..., stop-1
# range(start, stop, step) → start, start+step, ...

for i in range(5):          # 0 to 4
    print(i, end=" ")
print()

for i in range(2, 8):       # 2 to 7
    print(i, end=" ")
print()

for i in range(0, 20, 3):   # 0, 3, 6, 9, 12, 15, 18
    print(i, end=" ")
print()

for i in range(10, 0, -2):  # 10, 8, 6, 4, 2 (counting down!)
    print(i, end=" ")
print()
````

````python
# === enumerate() — get both index AND value ===
fruits = ["Mango", "Apple", "Banana", "Grapes"]

# Without enumerate (manual index)
for i in range(len(fruits)):
    print(f"{i}: {fruits[i]}")

print("---")

# With enumerate (cleaner!) ✅
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")

print("---")

# Start counting from 1
for index, fruit in enumerate(fruits, start=1):
    print(f"{index}. {fruit}")
````

## 3.3 `while` Loops ⏳

A `while` loop runs **as long as** a condition is `True`. Be careful — forgetting to update the condition leads to an **infinite loop**!

````python
# Basic while loop
count = 5
while count > 0:
    print(f"{count}...", end=" ")
    count -= 1          # DON'T forget this!
print("🚀 Liftoff!")
````

````python
# While with user input — keep asking until valid
# (In Jupyter, this will ask for input)
while True:
    answer = input("Guess the number (1-10): ")
    if answer == "7":
        print("🎉 Correct!")
        break                # exit the loop
    else:
        print("Try again!")
````

## 3.4 `break`, `continue`, `pass` 🎮

| Statement | What it does |
|-----------|-------------|
| `break` | Exit the loop immediately |
| `continue` | Skip the rest of this iteration, go to next |
| `pass` | Do nothing (placeholder) |

````python
# break — stop when found
print("=== break ===")
for num in range(1, 100):
    if num * num > 50:
        print(f"First number whose square > 50: {num} ({num*num})")
        break

# continue — skip certain items
print("\n=== continue ===")
for num in range(1, 11):
    if num % 3 == 0:
        continue     # skip multiples of 3
    print(num, end=" ")  # 1 2 4 5 7 8 10
print()

# pass — placeholder for future code
print("\n=== pass ===")
for num in range(5):
    if num == 3:
        pass     # TODO: handle this case later
    else:
        print(num, end=" ")
print()
````

## 3.5 Nested Loops 🔄🔄

A loop inside another loop. The inner loop runs **completely** for each iteration of the outer loop.

````python
# How nested loops work
for i in range(1, 4):       # outer: 3 iterations
    for j in range(1, 4):   # inner: 3 iterations EACH time
        print(f"({i},{j})", end=" ")
    print()                 # new line after each outer iteration

# Total: 3 × 3 = 9 prints
````

## 3.6 Loop Patterns & Idioms 🧩

Common patterns you'll use over and over.

````python
numbers = [4, 8, 15, 16, 23, 42]

# === Pattern 1: Accumulator ===
total = 0
for n in numbers:
    total += n
print(f"Sum: {total}")

# === Pattern 2: Counter ===
even_count = 0
for n in numbers:
    if n % 2 == 0:
        even_count += 1
print(f"Even numbers: {even_count}")

# === Pattern 3: Search ===
target = 23
found = False
for n in numbers:
    if n == target:
        found = True
        break
print(f"Found {target}: {found}")

# === Pattern 4: Build a new list ===
doubled = []
for n in numbers:
    doubled.append(n * 2)
print(f"Doubled: {doubled}")

# === Pattern 5: Filter ===
big_nums = []
for n in numbers:
    if n > 15:
        big_nums.append(n)
print(f"Big numbers: {big_nums}")
````

## 🎯 Exercise 15 — Mini-Project: Number Guessing Game 🎲

Build a complete guessing game using: `while` loop, `if/elif/else`, `input()`, `break`, `continue`, counter, and `import random`.

````python
import random

# =============================================
#    🎯 MINI-PROJECT: Number Guessing Game
# =============================================

print("🎲 NUMBER GUESSING GAME")
print("=" * 35)
print("I'm thinking of a number between 1 and 100.")
print("You have 7 attempts. Let's go!\n")

# TODO: Generate a random number between 1 and 100
secret = random.randint(1, 100)
max_attempts = 7
attempts = 0
won = False

while attempts < max_attempts:
    remaining = max_attempts - attempts
    
    # TODO: Get user's guess
    guess_str = input(f"Attempt {attempts + 1}/{max_attempts} ({remaining} left) — Your guess: ")
    
    # TODO: Validate input (skip if not a number)
    if not guess_str.isdigit():
        print("  ⚠️ Please enter a number!")
        ___
    
    guess = int(guess_str)
    attempts += 1
    
    # TODO: Check the guess
    if guess == secret:
        won = True
        ___
    elif guess < secret:
        # TODO: Give a hint based on how far off they are
        diff = secret - guess
        if diff > 20:
            print("  📈 Way too low!")
        else:
            print("  📈 Too low, but getting warm!")
    else:
        diff = guess - secret
        if diff > 20:
            print("  📉 Way too high!")
        else:
            print("  📉 Too high, but getting warm!")

# TODO: Print result
print("\n" + "=" * 35)
if won:
    print(f"🎉 CORRECT! The number was {secret}!")
    print(f"🏆 You got it in {attempts} attempt{'s' if attempts != 1 else ''}!")
    # Rate performance
    if attempts <= 3:
        print("⭐⭐⭐ Amazing!")
    elif attempts <= 5:
        print("⭐⭐ Great job!")
    else:
        print("⭐ Close call!")
else:
    print(f"💀 Game Over! The number was {secret}.")
    print("Better luck next time!")
````
